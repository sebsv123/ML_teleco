# Feature Selection

```python
X = df.drop(columns='diagnosis')
y = df.diagnosis
```

```python
""" chi2 """

from sklearn.feature_selection import SelectKBest, chi2

selector = SelectKBest(score_func=chi2, k=10)
selector.fit(X, y)
selector.pvalues_
```

```python
(
    pd.DataFrame(zip(X.columns, selector.pvalues_), columns=['feat','pval'])
    .sort_values(by='pval', ascending=False)
    .head(5)
)
```

```python
print(X.shape)
X_new = selector.transform(X)
print(X_new.shape)
```

```python
""" PCA """

from sklearn.decomposition import PCA

pca = PCA(n_components=9)
pca.fit(X)
pca.explained_variance_ratio_
```

```python
print(X.shape)
X_new = pca.transform(X)
print(X_new.shape)
```

```python
# Ejercicio
# pipeline, chi2 (10), logistic regression,
# train-test split

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline

seed = 99

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=.2, random_state=seed)

for ii in range(1,10):
    
    selector = SelectKBest(score_func=chi2, k=ii)
    pca = PCA(n_components=ii)
    scaler = MinMaxScaler()
    alg = LogisticRegression(random_state=seed)
    
    pipe = make_pipeline(pca, alg)

    pipe.fit(X_train, y_train)
    acc = pipe.score(X_test, y_test)
    print(f"{ii}: {acc:.3f}")
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
```

```python
seed = 74

models = []
models.append(("LR", LogisticRegression(random_state=seed)))
models.append(("DTC", DecisionTreeClassifier(random_state=seed)))
models.append(("NB", GaussianNB()))
models.append(("KNN", KNeighborsClassifier()))
models.append(("LDA", LinearDiscriminantAnalysis()))
models.append(("RF", RandomForestClassifier(random_state=seed)))
models.append(("SVM", SVC(random_state=seed)))
```

```python
cv_technique = KFold(n_splits=10, shuffle=True, random_state=seed)

cv_results = []
names = []
for name, model in models:
    
    results = cross_val_score(model, X, y, cv=cv_technique, scoring='roc_auc')
    
    cv_results.append(results)
    names.append(name)
    
plt.figure(figsize=(8,4))
plt.boxplot(cv_results)
plt.xticks(range(1, len(names)+1), names)
plt.show()
```

# Model tuning

```python
from sklearn.model_selection import GridSearchCV

cv_technique = KFold(n_splits=10, shuffle=True, random_state=seed)
model = DecisionTreeClassifier()

params = {
    "criterion": ["gini", "entropy", "log_loss"],
    "max_depth": [3,6,12,25,50],
    "max_features": [2,6,10],
    "random_state": [seed]
}

grid_model = GridSearchCV(estimator=model, cv=cv_technique, param_grid=params, scoring='accuracy')
grid_model.fit(X, y)
```
